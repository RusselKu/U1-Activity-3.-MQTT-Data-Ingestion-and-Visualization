```python
import json
import random
import ssl
import time
import paho.mqtt.client as mqtt

BROKER = "bird.lmq.cloudamqp.com"
PORT = 8883
USERNAME = "ygvefxav"
PASSWORD = "7IP9KbugtgqrlgcgNXo4KXy65mpaRNnn"

TOPIC_INT = "lake/raw/int"
TOPIC_FLOAT = "lake/raw/float"

def on_connect(client, userdata, flags, rc):
    print(f"[CONNECT] rc={rc}")

client = mqtt.Client(protocol=mqtt.MQTTv311)
client.username_pw_set(USERNAME, PASSWORD)

client.tls_set(
    tls_version=ssl.PROTOCOL_TLSv1_2,
    cert_reqs=ssl.CERT_NONE
)
client.tls_insecure_set(True)

client.on_connect = on_connect
client.connect(BROKER, PORT, 60)
client.loop_start()

try:
    while True:
        int_value = random.randint(0, 1000)
        float_value = round(random.uniform(0, 100), 4)

        payload_int = json.dumps({"value": int_value})
        payload_float = json.dumps({"value": float_value})

        client.publish(TOPIC_INT, payload_int, qos=0)
        client.publish(TOPIC_FLOAT, payload_float, qos=0)

        print(f"[PUBLISH] {TOPIC_INT} -> {payload_int}")
        print(f"[PUBLISH] {TOPIC_FLOAT} -> {payload_float}")

        time.sleep(2)

except KeyboardInterrupt:
    print("\n[STOP] Publisher stopped")

finally:
    client.loop_stop()
    client.disconnect()
```
